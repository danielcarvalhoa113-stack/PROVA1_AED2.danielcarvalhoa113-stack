R= acredito que a estrutura de fila seja o mais ideal para os atendimentos pois segue um ordem de chegada, e os pedidos já atendidos iriam ficar melhor na pilha pois é necessário o cancelamento

R= primeiro serão adicionados 4 chamados, os dois primeiros (Ana e Bruno) serão retirados da fila e colocados na pilha de já atendidos. O chamado do bruno vai ser desfeito, ele será retirado da pilha e será jogado no final da filha de atendimento. Por fim, o chamado da Carla será atendido e incluso nos chamados atendidos.
Assim vai ficar a ordem: Chamados a ser atendidos(Daniel e depois Bruno) Chamados já atendidos(Ana e depois Carla)

```python
# Classe de atendimento
class CentralAtendimento:
    def __init__(self):
        self.codigo = codigo
        self.cliente = cliente
        self.prioridade = prioridade
        self.chamado_espera = []
        self.chamado_atendido = []

        # Estrutura para chamados não atendidos
    def adicionar_chamado(self, codigo, cliente, prioridade):
        self.chamado_espera.append(self.codigo = [self.cliente, self.prioridade])

        # Atendendo um chamado
    def atender_proximo(self):
        self.atendimento = self.chamado_espera(0)
        self.chamado_atendido.append(self.atendimento)
        self.chamado_espera.remove(0)

        # Desfazer último atendimento
    def desfazer_ultimo_atendimento(self):
        self.anulado = self.chamado_atendido(-1)
        self.chamado_espera.append(self.anulado)
        self.chamado_atendido.pop()

        # Buscar chamado por código
    def buscar_chamado(self, codigo):
        for i in self.chamado_espera:
            if i == codigo:
                return i
            else:
                print(f"chamado não encontrado")

        # Exibir situação
    def exibir_situação(self):
        print(f"Chamados em aguardo")
        print({self.chamado_espera})
        print(f"Chamados atendidos")
        print({self.chamado_atendido})
```

```python
central = CentralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")
```
